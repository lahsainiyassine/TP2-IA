```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. Création de 3 phrases différentes (sport et cinéma)
# Astuce : certains mots sont partagés ('match', 'magnifique') pour bien observer la matrice
corpus = [
    "Le joueur marque un but magnifique pendant le match de football",
    "Le match de football est très intense et disputé",
    "Le film au cinéma raconte une histoire magnifique"
]

# 2. Initialisation et apprentissage du modèle CountVectorizer
# On utilise les stop words français pour retirer les articles simples
vectorizer_bow = CountVectorizer()
X_bow = vectorizer_bow.fit_transform(corpus)

# 3. Extraction du vocabulaire (liste des mots uniques)
vocabulaire = vectorizer_bow.get_feature_names_out()

# 4. Conversion en DataFrame pandas pour un affichage lisible
df_bow = pd.DataFrame(X_bow.toarray(), columns=vocabulaire, index=["Phrase 1", "Phrase 2", "Phrase 3"])

print("--- Liste des mots extraits (Vocabulaire) ---")
print(vocabulaire)
print("\n--- Matrice Bag of Words ---")
display(df_bow)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Utilisation du même corpus que la Partie 5
# corpus = [
#     "Le joueur marque un but magnifique pendant le match de football",
#     "Le match de football est très intense et disputé",
#     "Le film au cinéma raconte une histoire magnifique"
# ]

# 2. Initialisation et calcul de TF-IDF
vectorizer_tfidf = TfidfVectorizer()
X_tfidf = vectorizer_tfidf.fit_transform(corpus)

# 3. Récupération des mots et création du DataFrame
vocabulaire_tfidf = vectorizer_tfidf.get_feature_names_out()
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=vocabulaire_tfidf, index=["Phrase 1", "Phrase 2", "Phrase 3"])

# Arrondi à 2 décimales pour la clarté
print("--- Matrice TF-IDF (Pondérations) ---")
display(df_tfidf.round(3))
```
